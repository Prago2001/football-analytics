```python
from parser.parse_events import EventsDataParser
from parser.parse_stats import StatsParser
from scraper.match_scraper import ScrapeMatchData
import json
import pandas as pd
```

```python
def scrape_and_store(link: str):
    print(f"Scraping data for URL {link}")
    match_data = ScrapeMatchData()
    e, s = match_data.scrape(link)
    parser = EventsDataParser(e)
    df_events, df_qualifiers = parser.parse_events_and_qualifiers()
    df_metadata = parser.create_metadata_dataframe()
    EventsDataParser.load_or_append_events(df_events)
    EventsDataParser.load_or_append_metadata(df_metadata)
    EventsDataParser.load_or_append_qualifiers(df_qualifiers)
    df_stats = StatsParser(s).parse()
    stats_df = StatsParser.load_and_save_stats(df_stats, append=True)
    print(f"Finished parsing data for URL {link}")
```

```python
scrape_and_store("https://theanalyst.com/opta-football-match-centre?competitionId=2kwbbcootiqqgmrzs6o5inle5&seasonId=51r6ph2woavlbbpk8f29nynf8&matchId=3mupduyj99i9jq4xrg9rneoes")
```
